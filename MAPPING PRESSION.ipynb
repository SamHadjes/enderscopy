## Enderlights demo

This notebook shows how to connect and drive an enderscope illumination from Python.

```python
import time
from enderscope import SerialUtils, Balance, SerialDevice
b = Balance('/dev/serial/by-id/usb-Prolific_Technology_Inc._USB-Serial_Controller_D-if00-port0',baud_rate=9600)
print (b)
```

```python
def mesure_pression():
    # querrying LED state and example of how to send low level cmd 
    done = b.write(f"w")
    done = [b.decode('utf8') for b in done.split()]
    return float(done[0])
mesure_pression()
```

```python
from enderscope import ScanPatterns

origin = [80,120] # coordinates of first position
steps = [5,-5] # x and y shifts to next item
positions = ScanPatterns.raster(4,4) * steps + origin
#ScanPatterns.plot_path(positions, field=(0,0),title='raster scan')
```

```python
from enderscope import Stage
s = Stage('/dev/serial/by-id/usb-1a86_USB_Serial-if00-port0', 115200)
```

```python
s.home()
s.move_relative(0,0,-10)
#s.write_code(f"G92 Z 30")
s.write_code(f"M211 S0")
```

```python
s.move_relative(0,0,-0.1)
```

```python
s.get_position()
```

```python
def z_profile(max=10):
    pos = s.get_position()
    #s.move_position((pos[0],pos[1],10))
    pression = mesure_pression()
    while pression < max:
        s.move_relative(0,0,-0.05)
        s.finish_moves()
        pression = mesure_pression()
    print (pression, s.get_position()[2])
    s.move_position(pos)
    s.finish_moves()
```

```python
z_profile(5000)
```

```python
for p in positions:
    s.move_position(p)
    s.finish_moves()
    z_profile()
```
